```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# import data
file_path = "country_vaccinations.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
# problem 2 a - sina plot:

# convert date column to alternate datetime format:
df["date"] = pd.to_datetime(df["date"])

# my three countries selected for plotting:
selected_countries = ["United States", "India", "Brazil"]

# filtering the dataset for my selected countries and drop NaN values:
df_filtered = df[df["country"].isin(selected_countries)][["country", "total_vaccinations"]].dropna()

# plotting the sina plot (violin plot overlaid with individual data points):
sns.set(style = "whitegrid")
plt.figure(figsize=(10, 6))
sns.violinplot(x="country", y="total_vaccinations", data=df_filtered, inner=None, color="lightgray")
sns.stripplot(x="country", y="total_vaccinations", data=df_filtered, jitter=True, alpha=0.5)

plt.xlabel("Country")
plt.ylabel("Total Vaccinations per Country (Log Scale)")
plt.title("Sina Plot of Total Vaccinations for Selected Countries")
plt.show()
```

```python
# problem 2b - line plot of vaccinations in countries over time

# filtering the dataset for the line plot:
df_time_series = df[df["country"].isin(selected_countries)][["country", "date", "total_vaccinations"]].dropna()

# line plot for vaccination progress over time
sns.set(style="whitegrid")
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_time_series, x="date", y="total_vaccinations", hue="country")


plt.yscale("log")  # using log scale for better visualization
plt.xlabel("Date")
plt.ylabel("Total Vaccinations (Log Scale)")
plt.title("Progression of Total Vaccinations Over Time")
plt.legend(title="Country")
plt.xticks(rotation=45)
plt.show()
```
